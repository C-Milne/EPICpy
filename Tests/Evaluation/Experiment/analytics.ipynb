```python
import matplotlib.pyplot as plt
import pandas as pd

```

# Ingest Data

```python
# hamming = pd.read_csv('Hamming-Distance-seen-states-results.csv') # Basic hamming
hamming_newest = pd.read_csv('Hamming-Distance-seen-states-newest-first-results.csv') # Hamming with LIFO tie breaking
novelty_facts1 = pd.read_csv('Novelty_Facts_Only_Task-results.csv') # Basic novelty level 1
novelty_facts2 = pd.read_csv('Novelty_level2_Task-Method-Expand-0-TreeDis-results.csv') # Novelty for New facts and pairs of facts
novelty_methods = pd.read_csv('Novelty_Facts_Only_Task-Method-Expand-0-TreeDis-results.csv') # Novelty for method instance
bfs = pd.read_csv('seen-state-breadth-first-results.csv')
tree_distance = pd.read_csv('Tree-Distance-seen-states-newest-first-results.csv') # Basic heuristics 
# tree_distance = pd.read_csv('Tree-Distance-seen-states-results.csv') # Basic heuristics 

methods = {}
methods['Hamming'] = hamming_newest
methods['Novelty 1'] = novelty_facts1
methods['Novelty 2'] = novelty_facts2
methods['Novelty Methods'] = novelty_methods
# methods['BFS'] = bfs
methods['Tree Distance'] = tree_distance

domains = [
    'Rover',
    'Barman',
    'Depots',
    'Factories',
    'AssemblyHierarchical',
    'Blocksworld-GTOHP',
    'Childsnack',
    'Elevator-Learned-ECAI-16',
    'Entertainment',
    'Freecell-Learned-ECAI-16',
]
```

```python
def filter_domain(data: pd.DataFrame, domain: str):
    return data[data['Problem'].str.startswith(domain)]

def plot_attribute(attribute: str, domains: list, methods: dict):
    for domain in domains:
        for method, data in methods.items():
            print('Plotting %s for domain %s'%(method, domain))
            y = filter_domain(data, domain)[attribute]
            if y is not None: plt.plot(y, label=method)
        plt.legend()
        plt.title(attribute+":"+domain)
        plt.show()
```
